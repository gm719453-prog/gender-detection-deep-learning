# Gender Detection – Dataset Exploration

This notebook explores the UTKFace dataset, visualizes class distributions, and previews sample images.

```python
import os
import sys
from pathlib import Path

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image

# Add project root to path
sys.path.insert(0, str(Path.cwd().parent))
from config import DATASET_RAW_DIR, CLASS_NAMES
from preprocess import parse_dataset

plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette('husl')
```

## 1. Dataset Overview

```python
df = parse_dataset(DATASET_RAW_DIR)
print(f'Total images: {len(df)}')
print(f'\nGender distribution:')
print(df['gender'].value_counts().to_string())
print(f'\nAge statistics:')
print(df['age'].describe())
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Gender distribution
gender_counts = df['gender'].value_counts()
axes[0].bar(['Male', 'Female'], gender_counts.values, color=['#4facfe', '#f093fb'], edgecolor='white', linewidth=2)
axes[0].set_title('Gender Distribution', fontsize=14, fontweight='bold')
axes[0].set_ylabel('Count')
for i, v in enumerate(gender_counts.values):
    axes[0].text(i, v + 200, str(v), ha='center', fontweight='bold')

# Age distribution
axes[1].hist(df['age'], bins=40, color='#667eea', edgecolor='white', alpha=0.8)
axes[1].set_title('Age Distribution', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Age')
axes[1].set_ylabel('Count')

plt.tight_layout()
plt.savefig('dataset_overview.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 2. Sample Images

```python
# Display sample images
fig, axes = plt.subplots(3, 6, figsize=(15, 8))
axes = axes.flatten()

sample_df = df.sample(18, random_state=42).reset_index(drop=True)

for idx, row in sample_df.iterrows():
    try:
        img = Image.open(row['filepath'])
        img = img.resize((100, 100))
        axes[idx].imshow(img)
        gender_label = 'Male' if row['gender'] == 0 else 'Female'
        axes[idx].set_title(f"Age: {row['age']}\n{gender_label}", fontsize=9)
    except:
        axes[idx].axis('off')
    axes[idx].axis('off')

plt.suptitle('Sample UTKFace Images', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('sample_images.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 3. Ethnicity Distribution

```python
ethnicity_map = {0: 'White', 1: 'Black', 2: 'Asian', 3: 'Indian', 4: 'Other'}
df['ethnicity_name'] = df['ethnicity'].map(ethnicity_map)

eth_counts = df['ethnicity_name'].value_counts()

fig, ax = plt.subplots(figsize=(8, 5))
colors = ['#4facfe', '#00f2fe', '#f093fb', '#f5576c', '#ffd700']
bars = ax.bar(range(len(eth_counts)), eth_counts.values, color=colors[:len(eth_counts)], edgecolor='white', linewidth=2)
ax.set_xticks(range(len(eth_counts)))
ax.set_xticklabels(eth_counts.index, rotation=45, ha='right')
ax.set_title('Ethnicity Distribution', fontsize=14, fontweight='bold')
ax.set_ylabel('Count')

for bar, val in zip(bars, eth_counts.values):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 200, str(val), 
            ha='center', fontweight='bold')

plt.tight_layout()
plt.savefig('ethnicity_distribution.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 4. Gender by Ethnicity

```python
cross = pd.crosstab(df['ethnicity_name'], df['gender'])
cross.columns = ['Male', 'Female']

fig, ax = plt.subplots(figsize=(8, 5))
cross.plot(kind='bar', ax=ax, color=['#4facfe', '#f093fb'], edgecolor='white', linewidth=1.5)
ax.set_title('Gender Distribution by Ethnicity', fontsize=14, fontweight='bold')
ax.set_xlabel('Ethnicity')
ax.set_ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Gender')

plt.tight_layout()
plt.savefig('gender_by_ethnicity.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 5. Summary Statistics

```python
print('=' * 50)
print('  DATASET SUMMARY')
print('=' * 50)
print(f'Total images:     {len(df):>8}')
print(f'Male images:      {(df["gender"]==0).sum():>8}')
print(f'Female images:    {(df["gender"]==1).sum():>8}')
print(f'Balance ratio:    {(df["gender"]==1).sum()/(df["gender"]==0).sum():>8.3f}')
print(f'Mean age:         {df["age"].mean():>8.1f}')
print(f'Median age:       {df["age"].median():>8.0f}')
print(f'Age range:        {df["age"].min()} – {df["age"].max()}')
```
